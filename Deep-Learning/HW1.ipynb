Problem 1 - Multivariate Linear Regression Model and Loss

In this problem, we formulate a multivariate linear regression model using matrix notation. The bias/intercept term is incorporated into the weight matrix by adding a column of ones to the input matrix.




1.1 Model Notation

Let: 
- n be the number of training examples
- d be the number of input features
- q be the number of output variables 

Let $X_{\text{original}}$ be the input matrix. 

X =
\begin{bmatrix}
1 & x_{11} & x_{12} & \cdots & x_{1p}\\
1 & x_{21} & x_{22} & \cdots & x_{2p}\\
\vdots & \vdots & \vdots & \ddots & \vdots\\
1 & x_{n1} & x_{n2} & \cdots & x_{np}
\end{bmatrix}.

Therefore,

$X \in \mathbb{R}^{n\times(d+1)}$.

Let Y be the matrix of target outputs:

$Y \in \mathbb{R}^{n\times q}$.

Because the model has q output variables, the weight matrix is:

$W \in \mathbb{R}^{(d+1)\times q}$.




1.2 Model Prediction

The predicted output matrix is given by


$\hat{Y} = XW$.


The dimensions are


$X \in \mathbb{R}^{n\times(d+1)}$


and


$W \in \mathbb{R}^{(d+1)\times q}$.


Therefore, the resulting prediction matrix has dimensions


$\hat{Y} \in \mathbb{R}^{n\times q}$.


The matrix multiplication is valid because the inner dimensions match:


$(n\times(d+1))((d+1)\times q)=n\times q$.





1.3 Mean Squared Error Loss

The mean squared error (MSE) measures the average squared difference between the actual outputs Y and the predicted outputs $\hat{Y}$.

For n observations and q output variables, the MSE loss is

L(W)=$\frac{1}{nq}\sum_{i=1}^{n}\sum_{j=1}^{q}
\left(Y_{ij}-\hat{Y}_{ij}\right)^2$.

Using the prediction equation $\hat{Y}=XW$, this can also be written as

L(W)=$\frac{1}{nq}\sum_{i=1}^{n}\sum_{j=1}^{q}
\left(Y_{ij}-(XW)_{ij}\right)^2$.

The error matrix is

E = Y-$\hat{Y}$ = Y-XW,

with dimensions

$E\in\mathbb{R}^{n\times q}$.

The MSE loss is a scalar value:

$L(W)\in\mathbb{R}$.
